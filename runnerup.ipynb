```python
scores = [10, 20, 4, 45, 99, 99, 50]

highest = float('-inf')
second_highest = float('-inf')

for score in scores:
    if score > highest:
        second_highest = highest
        highest = score
    elif highest > score > second_highest:
        second_highest = score

print("Second-highest score:", second_highest)
```
